```python
df = df.drop_duplicates()
```

```python
df = df[
    [
        "product_name",
        "brand_name",
        "rating",
        "price_usd",
        "ingredients",
        "primary_category",
        "secondary_category",
        "loves_count"
    ]
]
```

```python
df = df.dropna(
    subset=[
        "primary_category",
        "rating"
    ]
)
```

```python
df = df.reset_index(drop=True)
```

```python
df.to_csv(
    "../data/processed/clean_products.csv",
    index=False
)
```

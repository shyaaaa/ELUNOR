```python
import pandas as pd

from sklearn.cluster import KMeans

from sklearn.preprocessing import StandardScaler

import matplotlib.pyplot as plt
```

```python
df = pd.read_csv(
    "../data/processed/clean_products.csv"
)
```

```python
features = df[
    [
        "rating",
        "price_usd",
        "loves_count"
    ]
]
```

```python
features = features.dropna()
```

```python
scaler = StandardScaler()

scaled_features = scaler.fit_transform(
    features
)
```

```python
kmeans = KMeans(
    n_clusters=4,
    random_state=42
)
```

```python
kmeans.fit(scaled_features)
```

```python
features["cluster"] = kmeans.labels_
```

```python
print(
    features["cluster"]
    .value_counts()
)
```

```python
plt.figure(figsize=(10,6))

plt.scatter(
    features["price_usd"],
    features["loves_count"],
    c=features["cluster"]
)

plt.xlabel("Price USD")

plt.ylabel("Loves Count")

plt.title("Product Segmentation")

plt.show()
```

```python
cluster_analysis = (
    features
    .groupby("cluster")
    .mean()
)

print(cluster_analysis)
```
